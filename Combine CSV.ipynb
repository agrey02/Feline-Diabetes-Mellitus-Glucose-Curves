```python
# Combining CSV files of cat blood glucose and insulin data

import pandas as pd
from pathlib import Path

data_folder = Path("Cat_CSV")
cats = []
```

```python
for file in data_folder.glob("*.csv"):

    # selecting specific data range from raw data files
    cat_data = pd.read_csv(file, skiprows = 3, usecols = range(27), dtype = str)
    
    # assigning identification for each of the insulin curves
    cat_data["ID"] = file.stem
    cat_data["source"] = file.name

    cats.append(cat_data)
    
    master_data = pd.concat(cats, ignore_index = True)
    master_data.to_csv("master_data.csv", index = False) 
    
```
